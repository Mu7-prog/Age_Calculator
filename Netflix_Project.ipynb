Replacing missing directors element with popular director from that relevant country

```python
director_popularity = (
    netflix_df[~netflix_df['director'].isna()]
    .groupby('country')['director']
    .agg(lambda x: x.value_counts().idxmax())
    .to_dict()
)

netflix_df['popular_director'] = netflix_df['country'].map(director_popularity)
netflix_df['director'] = netflix_df['director'].fillna(netflix_df['popular_director'])
netflix_df.drop(columns=['popular_director'], inplace=True)

cast_popularity = (
    netflix_df[~netflix_df['cast'].isna()]
    .groupby('country')['cast']
    .agg(lambda x: x.value_counts().idxmax())
    .to_dict()
)

netflix_df['popular_cast'] = netflix_df['country'].map(cast_popularity)
netflix_df['cast'] = netflix_df['cast'].fillna(netflix_df['popular_cast'])
netflix_df.drop(columns=['popular_cast'], inplace=True)

country_popularity = (
    netflix_df[~netflix_df['country'].isna()]
    .groupby('cast')['country']
    .agg(lambda x: x.value_counts().idxmax())
    .to_dict()
)

netflix_df['popular_country'] = netflix_df['cast'].map(country_popularity)
netflix_df['country'] = netflix_df['country'].fillna(netflix_df['popular_country'])
netflix_df.drop(columns=['popular_country'], inplace=True)

netflix_df.loc[
    netflix_df['country'].isna() & netflix_df['director'].isna() & netflix_df['cast'].isna(),
    ['country', 'director', 'cast']
] = ['No country info', 'No director info', 'No cast info']

netflix_df.loc[
    netflix_df['country'].isna() & netflix_df['director'].isna(),
    ['country', 'director']
] = ['No country info', 'No director info']

netflix_df.loc[
    netflix_df['country'].isna() & netflix_df['cast'].isna(),
    ['country', 'cast']
] = ['No country info', 'No cast info']

netflix_df.loc[
    netflix_df['director'].isna() & netflix_df['cast'].isna(),
    ['director', 'cast']
] = ['No director info', 'No cast info']

netflix_df.loc[netflix_df['country'].isna(), 'country'] = 'No country info'
netflix_df.loc[netflix_df['director'].isna(), 'director'] = 'No director info'
netflix_df.loc[netflix_df['cast'].isna(), 'cast'] = 'No cast info'


```

Most movies directed by a director according to top 10 genres

```python
movies_df = netflix_df[netflix_df['type'] == 'Movie']
movies_df = movies_df.assign(genre=movies_df['listed_in'].str.split(', '))
movies_df = movies_df.explode('genre').dropna(subset=['genre'])

top_genres = (
    movies_df['genre']
    .value_counts()
    .head(10)
    .index
)
top_genre_movies = movies_df[movies_df['genre'].isin(top_genres)]
director_genre_counts = (
    top_genre_movies.groupby(['genre', 'director'])
    .size()
    .reset_index(name='movie_count')
    .sort_values(by=['genre', 'movie_count'], ascending=[True, False])
)
top_directors_by_genre = (
    director_genre_counts.groupby('genre')
    .head(1)
    .reset_index(drop=True)
)
plt.figure(figsize=(12, 8))
sns.barplot(
    data=top_directors_by_genre,
    x='movie_count',
    y=top_directors_by_genre.apply(lambda x: f"{x['genre']} ({x['director']})", axis=1),
    palette='viridis'
)
plt.title('Top Directors by Movie Count in Top 10 Genres', fontsize=16)
plt.xlabel('Number of Movies Directed', fontsize=12)
plt.ylabel('Genre (Director)', fontsize=12)
plt.xticks(fontsize=10)
plt.yticks(fontsize=10)
plt.tight_layout()
plt.show()
```

Outlier check

```python
numeric_columns = netflix_df.select_dtypes(include=[np.number]).columns
for column in numeric_columns:

    Q1 = netflix_df[column].quantile(0.25)
    Q3 = netflix_df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = netflix_df[(netflix_df[column] < lower_bound) | (netflix_df[column] > upper_bound)]

    print(f"\nColumn: {column}")
    print(f"Number of Outliers: {len(outliers)}")
    if len(outliers) > 0:
        print(outliers[[column]].describe())


    plt.figure(figsize=(8, 4))
    sns.boxplot(data=netflix_df, x=column)
    plt.title(f'Boxplot for {column} (Outlier Detection)')
    plt.show()
```
